````python
bins = [0, 18, 25, 40, 60, 100]
labels = ['<18', '18-25', '26-40', '41-60', '60+']
customers_df['age_group'] = pd.cut(customers_df['age'], bins=bins, labels=labels, right=False)

display(customers_df.head())
````

````python

order_df['order_date'] = pd.to_datetime(order_df['order_date'])
order_df['order_month'] = order_df['order_date'].dt.to_period('M')


display(order_df.head())
````

````python

customer_order_counts = order_df['customer_id'].value_counts()
repeat_customers = customer_order_counts[customer_order_counts > 1].index
order_df['is_repeat_customer'] = order_df['customer_id'].isin(repeat_customers)
display(order_df.head())
display(order_df[order_df['is_repeat_customer'] == True].head())
````

````python
order_df['total_amount'] = order_df['quantity'] * order_df['price']
````

### Step 4 — Groupby Analytics

````python
merged_df = pd.merge(order_df, customers_df, on='customer_id', how='left')
merged_df_1=pd.merge(merged_df,products_df,on='product_id',how='left')
display(merged_df_1.head())
````

````python
revenue_per_city = merged_df_1.groupby('city')['total_amount'].sum().sort_values(ascending=False)
revenue_per_city
````

````python
monthly_revenue_trend = merged_df_1.groupby('order_month')['total_amount'].sum().reset_index()
monthly_revenue_trend['order_month'] = monthly_revenue_trend['order_month'].astype(str) # Convert Period to string for better display
display(monthly_revenue_trend)
````
