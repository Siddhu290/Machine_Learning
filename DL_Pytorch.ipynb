This cell loads the well-known Iris dataset from `sklearn.datasets`, converts it into a pandas DataFrame, adds the 'target' column, and then displays the first 5 rows of the DataFrame. This is a common step for initial data inspection.

```python
from sklearn.datasets import load_iris
import pandas as pd
# Load the Iris dataset
iris = load_iris()

# Create a DataFrame from the dataset
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
iris_df['target'] = iris.target

# Display the first 5 rows of the Iris dataset
display(iris_df.head())
```

Here, we define the feature columns for our model and extract them from the `iris_df` DataFrame, converting them into a NumPy array `x`. These will be the input features for our neural network.

```python
# Define feature columns from the Iris dataset
feature_columns = ['sepal length (cm)', 'sepal width (cm)', 'petal length (cm)', 'petal width (cm)']
x = iris_df[feature_columns].values
```

This cell extracts the 'target' column from the `iris_df` DataFrame, converting it into a NumPy array `y`. These are the ground-truth labels our model will try to predict.

```python
y = iris_df['target'].values
```

This cell creates a `TensorDataset` by pairing the input features `x` and target labels `y`. Both are converted to PyTorch tensors and cast to `float` data type, which is typically required for neural network inputs.

```python
dataset = TensorDataset(torch.tensor(x).float(), torch.tensor(y).float())
```

This cell displays the `TensorDataset` object, showing its type and memory location.

```python
dataset
```

This cell defines a batch size `n` (which is 32) and then creates a `DataLoader`. The `DataLoader` shuffles the dataset and provides an iterable that yields data in batches, which is crucial for efficient training of neural networks.

```python
n = 32 # Define batch size
dataloader = DataLoader(dataset, batch_size=n, shuffle=True)
```

```python
F.one_hot(torch.tensor([0,1,2]),num_classes=3)
```

This cell initializes a `Dropout` layer with a dropout probability of 0.5. Dropout is a regularization technique that randomly sets a fraction of input units to zero at each update during training, which helps prevent overfitting.

```python
nn.Dropout(p=0.5)
```

This cell defines the full neural network model using `nn.Sequential`. It consists of two hidden layers with ReLU activations, and an output layer with a Softmax activation for multi-class classification. The input features, hidden layer sizes, and number of output classes are determined dynamically from the data.

```python
from torch.nn.modules.activation import Softmax

# Define the input, hidden, and output sizes for the model
n_feature = x.shape[1]  # Number of input features (from 'x')
n_classes = len(torch.unique(torch.tensor(y))) # Number of output classes (from 'y')
i = 10  # Number of neurons in the first hidden layer
j = 8   # Number of neurons in the second hidden layer

model = nn.Sequential(nn.Linear(n_feature, i),
                       nn.ReLU(),
                       nn.Linear(i, j),
                       nn.ReLU(),
                       nn.Linear(j, n_classes),
                       nn.Softmax(dim=-1))
```

This cell performs a forward pass through the `model` with `sample_input` (which is our `x` features converted to a tensor). The output `prediction` represents the model's predicted probabilities for each class for every input sample.

```python
sample_input = torch.tensor(x).float()
prediction = model(sample_input)
print(prediction)
```

This cell converts the NumPy array `y` (target labels) into a PyTorch tensor and casts it to `long` data type. Integer labels are often required by loss functions like `CrossEntropyLoss`.

```python
actual = torch.tensor(y).long()
```

Here, the `actual` target labels are one-hot encoded using `F.one_hot` to match the dimensionality of the `prediction` tensor. Then, the Mean Squared Error (MSE) loss is calculated between the `prediction` and `one_hot_actual`.

```python
one_hot_actual = F.one_hot(actual, num_classes=n_classes).float()
mse_loss = nn.MSELoss()(prediction, one_hot_actual)
```

This cell calculates the Smooth L1 Loss between the model's `prediction` and the `one_hot_actual` target labels. Smooth L1 Loss is less sensitive to outliers than MSE Loss.

```python
l1_loss = nn.SmoothL1Loss()(prediction, one_hot_actual)
```

This cell calculates the Binary Cross-Entropy (BCE) loss between the model's `prediction` (probabilities) and the `one_hot_actual` target labels. BCE is typically used for binary classification, or multi-label classification where each label is independent.

```python
bc3_loss = nn.BCELoss()(prediction, one_hot_actual)
```

This cell calculates the Cross-Entropy (CE) loss between the model's `prediction` and the `one_hot_actual` target labels. Note that `nn.CrossEntropyLoss` typically expects raw logits (scores before Softmax) and integer labels, so using it with one-hot encoded probabilities might not be the standard use case here.

```python
ce_loss = nn.CrossEntropyLoss()(prediction, one_hot_actual)
```

This cell assigns the `mse_loss` to a variable named `loss` and then calls `loss.backward()`. This crucial step computes the gradients of the `loss` with respect to all learnable parameters in the model.

```python
loss = mse_loss
loss.backward()
```

This cell displays the value of the `loss` tensor. Note that `loss.backward()` only computes gradients; it does not change the `loss` value itself.

```python
loss
```

This cell initializes an `SGD` (Stochastic Gradient Descent) optimizer. It takes the model's learnable parameters (`model.parameters()`) and configures the learning rate (`lr=0.01`) and momentum (`momentum=0.95`), which are hyperparameters for the optimization process.

```python
from torch.optim import optimizer
optimizer = optim.SGD(model.parameters(), lr=0.01, momentum=0.95)
```

This cell displays the initialized `SGD` optimizer object, showing its configuration and the parameter groups it manages.

```python
optimizer
```

This cell calls `optimizer.step()`. This function performs a single optimization step, using the gradients computed by `loss.backward()` to update the model's parameters according to the chosen optimization algorithm (SGD in this case).

```python
optimizer.step()
```

This cell sets the model to training mode (`model.train()`). This is important because certain layers, like Dropout and BatchNorm, behave differently during training and evaluation.

```python
model.train()
```

This cell re-defines the loss criterion as `nn.MSELoss()`. This loss function will be used to quantify the difference between the model's predictions and the true labels during training.

```python
loss_criterion = nn.MSELoss()
```

This cell re-initializes the `SGD` optimizer for the model's parameters with the same learning rate and momentum as before. This step is necessary if the model parameters were updated or reset.

```python
optimizer = optim.SGD(model.parameters(), lr=0.01, momentum=0.95)
```

This cell displays the re-initialized `SGD` optimizer object, confirming its current configuration.

```python
optimizer
```

This cell contains the core training loop for a single epoch. It iterates through the `dataloader` to get batches of features and targets, performs a forward pass, calculates the loss, performs backpropagation (`loss.backward()`), and updates the model's parameters (`optimizer.step()`). `optimizer.zero_grad()` is called at the beginning of each batch to clear previous gradients.

```python
for data in dataloader:
  optimizer.zero_grad()
  features, target = data
  predictions = model(features)
  one_hot_target = F.one_hot(target.long(), num_classes=n_classes).float() # One-hot encode the target
  loss = loss_criterion(predictions, one_hot_target)
  loss.backward()
  optimizer.step()
```

This cell sets the model to evaluation mode (`model.eval()`). In this mode, layers like Dropout are disabled, and BatchNorm layers use their accumulated running statistics instead of batch statistics, ensuring consistent behavior during inference.

```python
model.eval()
```

This cell initializes an `Accuracy` metric from `torchmetrics`. It's configured for multiclass classification (`task='multiclass'`) and specifies the number of classes (`num_classes=3`). This metric will be used to evaluate the model's performance.

```python
metric = torchmetrics.Accuracy(task='multiclass', num_classes=3)
```

This cell iterates through the `dataloader` to evaluate the model's accuracy. For each batch, it gets predictions from the model and updates the `metric` with the predictions and the true target labels (converted to `long`). Finally, `metric.compute()` calculates the overall accuracy.

```python
for i , data in enumerate(dataloader, 0):
  features, target = data
  predictions = model(features)
  # For accuracy, we need the original target labels (not one-hot encoded)
  # and the predicted class from the model output
  accuracy = metric(predictions, target.long())

accuracy = metric.compute()
accuracy
```

This cell displays the final computed accuracy value from the `torchmetrics` object.

```python
accuracy
```

This cell resets the internal state of the `torchmetrics` accuracy metric. This is important if you want to calculate accuracy for a new set of predictions without being influenced by previous computations.

```python
metric.reset()
```
